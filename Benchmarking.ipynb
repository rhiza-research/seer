```python
import pypsa
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
%matplotlib inline

import time

from synthseer import *
%load_ext autoreload
%autoreload 2
```

```python
nyears = np.arange(1, 4)

modular = [True, False]
```

```python
cases = pd.MultiIndex.from_product([nyears, modular], names=["nyears", "modular"])
results = pd.DataFrame(index=cases, columns=["time (s)"])
```

```python
for ny in nyears:
    for mod in modular: 
        input_data = synth_data(2025, nyears=ny, load_growth=0.2, excess_cap=1)
        model = create_model_from_data(input_data, voll=5000, modular_expansion=mod)

        t0 = time.time()
        model.optimize()
        t1 = time.time()
        results.loc[(ny, mod)] = t1 - t0
        
```

```python
results.loc[:, "time (m)"] = results.loc[:, "time (s)"] / 60.0
```

```python
plt.plot(rez_milp["nyears"], rez_milp["time (s)"], marker="o", label="MILP")
plt.plot(rez_lp["nyears"], rez_lp["time (s)"], marker="o", label="LP")
plt.set_xt

plt.xlabel("Number of years")
plt.ylabel("Solver runtime (s)")
plt.legend()
plt.show()
```
